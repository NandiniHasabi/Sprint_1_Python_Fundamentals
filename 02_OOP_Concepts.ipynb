# Notebook 2 – Object-Oriented Programming (OOP)

## Introduction

Object-Oriented Programming (OOP) is a programming approach where we organize programs using objects and classes.

A class defines the structure and behavior of an object.

An object is an instance of a class.

OOP helps us organize code, reuse code, and represent real-world entities in a simple way.

In this notebook, I will learn and implement important Python OOP concepts using simple examples.

## 1. Class

### What is a Class?

A class is a blueprint or template used to create objects.

It defines the data and functions that an object can have.

### Example

A `Student` class can describe a student by storing information such as name and age.

The class is like a blueprint, while the actual student created from it is an object.

```python
class Student:
    name = "Nandini"
    age = 23

print(Student.name)
print(Student.age)
```

### Code Explanation

- `class Student:` creates a class named `Student`.
- `name` and `age` are data defined inside the class.
- `Student.name` accesses the name from the class.
- `Student.age` accesses the age from the class.

This demonstrates how a class can define data that belongs to a particular type of object.

## 2. Object

### What is an Object?

An object is an instance of a class.

After creating a class, we can create objects from that class.

### Example

If `Student` is a class, then `student1` can be an object of the `Student` class.

```python
class Student:
    name = "Nandini"

student1 = Student()

print(student1.name)
```

###  Code Explanation

- `Student` is the class.
- `student1` is an object created from the `Student` class.
- `student1.name` accesses the data defined in the class.

So, a class is the blueprint and an object is the actual instance created from that blueprint.

## 3. Constructor (__init__)

### What is a Constructor?

A constructor is a special method that is automatically called when an object is created.

In Python, the constructor is usually written as `__init__()`.

It is commonly used to initialize the data of an object.

### Example

When creating a student object, we can automatically provide the student's name and age.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Nandini", 23)  

print(student1.name)
print(student1.age)
```

### Code Explanation

- `__init__()` runs automatically when the object is created.
- `self.name` stores the student's name.
- `self.age` stores the student's age.
- `"Nandini"` and `23` are passed while creating the object.

The constructor helps initialize an object's data.

## 4. Instance Variables

### What are Instance Variables?

Instance variables are variables that belong to a particular object.

Each object can have different values for its instance variables.

### Simple Example

Two students can have different names and ages.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Nandini", 23)
student2 = Student("Anu", 22)

print(student1.name, student1.age)
print(student2.name, student2.age)
```

### Code Explanation

- `self.name` and `self.age` are instance variables.
- `student1` has its own name and age.
- `student2` has different values.
- Each object maintains its own instance data.

## 5. Class Variables

### What are Class Variables?

A class variable is a variable shared by all objects of a class.

It is defined inside the class but outside the methods.

### Example

If all students belong to the same college, we can store the college name as a class variable.

```python
class Student:
    college = "BGMIT"

    def __init__(self, name):
        self.name = name

student1 = Student("Nandini")
student2 = Student("Anu")

print(student1.name, student1.college)
print(student2.name, student2.college)
```

### Code Explanation

- `college` is a class variable.
- It is shared by all Student objects.
- `name` is an instance variable because each student can have a different name.

## 6. Instance Methods

### What is an Instance Method?

An instance method is a method that works with the data of a particular object.

It takes `self` as its first parameter.

### Example

A student can have a method to display their details.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_details(self):
        print("Name:", self.name)
        print("Age:", self.age)

student1 = Student("Nandini", 23)

student1.display_details()
```

### Code Explanation

- `display_details()` is an instance method.
- `self` refers to the current object.
- The method accesses `self.name` and `self.age`.
- `student1.display_details()` calls the method for the `student1` object.

## 7. Static Methods

### What is a Static Method?

A static method is a method that does not depend on the object or class data.

It does not use `self` or `cls`.

We use `@staticmethod` to create a static method.

### Example

A calculator method that adds two numbers does not need any information from a Student object, so it can be a static method.

```python
class Student:
    @staticmethod
    def add_marks(mark1, mark2):
        return mark1 + mark2


result = Student.add_marks(80, 90)

print("Total marks:", result)
```

### Code Explanation

- `@staticmethod` makes the method a static method.
- `add_marks()` does not use `self` or `cls`.
- It only works with the values passed to it.
- We can call the method using the class name.

## 8. Class Methods

### What is a Class Method?

A class method is a method that works with class-level data.

It uses `cls` as its first parameter.

We use `@classmethod` to create a class method.

### Example

If the college name belongs to the whole Student class, a class method can be used to display or change that information.

```python
class Student:
    college = "BGMIT"

    @classmethod
    def show_college(cls):
        print("College:", cls.college)


Student.show_college()
```

### Code Explanation

- `@classmethod` creates a class method.
- `cls` refers to the class itself.
- `cls.college` accesses the class variable.
- We can call the method using the class name.

## 9. Properties (@property)

### What is a Property?

A property allows us to access a method like an attribute.

It is useful when we want to control how a value is accessed or changed.

We use the `@property` decorator.

### Example

We can create a property for a student's name and access it like a normal variable.

```python
class Student:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name


student1 = Student("Nandini")

print(student1.name)
```

### Code Explanation

- `_name` stores the actual value.
- `@property` allows us to access the `name()` method like an attribute.
- We use `student1.name` instead of `student1.name()`.
- Properties are useful when we want controlled access to object data.

## 10. Encapsulation

### What is Encapsulation?

Encapsulation means keeping data and the methods that work with that data together inside a class.

It also helps control direct access to internal data.

In Python, we commonly use `_` or `__` to indicate protected or private-style attributes.

### Example

A bank account should not allow anyone to directly change the balance.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(5000)

account.show_balance()
```

### Code Explanation

- `__balance` is a private-style attribute.
- It is kept inside the `BankAccount` class.
- `show_balance()` provides controlled access to the balance.
- This demonstrates the idea of encapsulation.

## 11. Abstraction

### What is Abstraction?

Abstraction means hiding unnecessary implementation details and showing only the important functionality.

In Python, we can use the `abc` module to create abstract classes.

An abstract class can define methods that child classes must implement.

### Example

A vehicle can have a common `start()` operation, but different vehicles can implement it differently.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts with a key")


class Bike(Vehicle):

    def start(self):
        print("Bike starts with a button")


car = Car()
bike = Bike()

car.start()
bike.start()
```

### Code Explanation

- `ABC` is used to create an abstract class.
- `@abstractmethod` defines a method that child classes must implement.
- `Vehicle` provides the common structure.
- `Car` and `Bike` provide their own implementation of `start()`.
- The user only needs to know that the vehicle can start; the internal implementation is different.

This demonstrates abstraction by hiding the implementation details.

## 12. Inheritance

### What is Inheritance?

Inheritance allows one class to use the properties and methods of another class.

The existing class is called the parent class, and the new class is called the child class.

Inheritance helps us reuse existing code instead of writing the same code again.

### Example

A `Dog` is an `Animal`, so the Dog class can inherit common behavior from the Animal class.

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()
dog.bark()
```

### Code Explanation

- `Animal` is the parent class.
- `Dog` is the child class.
- `Dog(Animal)` means Dog inherits from Animal.
- The Dog object can use both `eat()` and `bark()`.
- This demonstrates code reuse through inheritance.

## 13. Multiple Inheritance

### What is Multiple Inheritance?

Multiple inheritance means that one child class inherits from more than one parent class.

### Example

A person can have skills from both a Programmer class and a Designer class.

```python
class Programmer:
    def write_code(self):
        print("Writing Python code")


class Designer:
    def design_ui(self):
        print("Designing user interface")


class Developer(Programmer, Designer):
    pass


developer = Developer()

developer.write_code()
developer.design_ui()
```

### Code Explanation

- `Programmer` and `Designer` are two parent classes.
- `Developer` inherits from both classes.
- The Developer object can use methods from both parents.
- This is called multiple inheritance.

## 14. Multilevel Inheritance

### What is Multilevel Inheritance?

Multilevel inheritance occurs when a class inherits from another child class.

It creates a chain of inheritance.

Example:

Grandparent → Parent → Child

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Puppy(Dog):
    def play(self):
        print("Puppy is playing")


puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.play()
```

### Code Explanation

- `Animal` is the first-level parent.
- `Dog` inherits from `Animal`.
- `Puppy` inherits from `Dog`.
- The Puppy object can access methods from both Dog and Animal.
- This creates a multilevel inheritance chain.

## 15. Hierarchical Inheritance

### What is Hierarchical Inheritance?

Hierarchical inheritance occurs when multiple child classes inherit from the same parent class.

### Example

Different types of vehicles can inherit common functionality from a Vehicle class.

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Bike(Vehicle):
    def ride(self):
        print("Bike is riding")


car = Car()
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
```

### Code Explanation

- `Vehicle` is the parent class.
- `Car` and `Bike` are child classes.
- Both child classes inherit the `start()` method.
- Each child class also has its own method.
- This is called hierarchical inheritance.

## 16. Polymorphism

### What is Polymorphism?

Polymorphism means "many forms".

In Python, different objects can respond to the same method name in different ways.

### Example

Different animals can have a `sound()` method, but each animal produces a different sound.

```python
class Dog:
    def sound(self):
        print("Dog says: Woof")


class Cat:
    def sound(self):
        print("Cat says: Meow")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

### Code Explanation

- Both `Dog` and `Cat` have a method named `sound()`.
- The same method call `animal.sound()` produces different results.
- Dog gives one output and Cat gives another output.
- This demonstrates polymorphism.

## 17. Method Overriding

### What is Method Overriding?

Method overriding happens when a child class provides its own implementation of a method that already exists in the parent class.

### Example

A parent class can have a `sound()` method, and the child class can provide its own version of that method.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog says: Woof")


animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
```

### Code Explanation

- `Animal` has a `sound()` method.
- `Dog` inherits from `Animal`.
- `Dog` defines its own `sound()` method.
- The Dog version replaces the inherited implementation when called on a Dog object.
- This is called method overriding.

## 18. Method Overloading – Python Approach

### What is Method Overloading?

Method overloading means having multiple methods with the same name but different parameters.

Python does not support traditional method overloading like some other programming languages.

Instead, we can achieve similar behavior using default arguments or `*args`.

### Example

We can create one method that can add either two numbers or three numbers.

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

### Code Explanation

- The `add()` method has three parameters.
- `c` has a default value of `0`.
- When two values are passed, `c` uses its default value.
- When three values are passed, all three are added.
- This is one common way to achieve method-overloading-like behavior in Python.

## 19. Composition

### What is Composition?

Composition means creating a strong "has-a" relationship between objects.

One class contains an object of another class.

The contained object is usually created inside the containing class.

### Example

A `Car` has an `Engine`.

The Car class creates and owns its Engine object.

```python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")


car = Car()
car.start_car()
```

### Code Explanation

- `Engine` is a separate class.
- `Car` creates an `Engine` object inside its constructor.
- The Car uses the Engine through `self.engine`.
- This represents a strong "has-a" relationship.
- This is an example of composition.

## 20. Aggregation

### What is Aggregation?

Aggregation is also a "has-a" relationship, but the contained object can exist independently of the container.

The objects have a weaker relationship compared with composition.

### Example

A `Department` has teachers, but teachers can exist even if the department is removed.

```python
class Teacher:
    def __init__(self, name):
        self.name = name


class Department:
    def __init__(self, teacher):
        self.teacher = teacher

    def show_teacher(self):
        print("Teacher:", self.teacher.name)


teacher1 = Teacher("Anita")
department = Department(teacher1)

department.show_teacher()
```

### Code Explanation

- `Teacher` is created separately.
- The `Department` receives the existing Teacher object.
- The Teacher can exist independently of the Department.
- This represents aggregation.

## 21. Association

### What is Association?

Association means that two classes are connected or interact with each other.

There is no strong ownership between the objects.

### Example

A `Student` and a `Teacher` can interact with each other, but neither object owns the other.

```python
class Teacher:
    def teach(self):
        print("Teacher is teaching")


class Student:
    def learn(self, teacher):
        print("Student is learning")
        teacher.teach()


teacher = Teacher()
student = Student()

student.learn(teacher)
```

### Code Explanation

- `Teacher` and `Student` are separate classes.
- The Student interacts with the Teacher through the `learn()` method.
- Neither class owns the other.
- This represents association.

## 22. Magic / Dunder Methods

### What are Magic or Dunder Methods?

Magic methods are special methods in Python whose names start and end with double underscores.

For example:

- `__init__()`
- `__str__()`
- `__len__()`
- `__repr__()`

They allow Python objects to work with built-in operations and functions.

### Example

We can define `__len__()` to tell Python how to calculate the length of our object.

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Nandini", "Anu", "Priya"])

print("Number of members:", len(team))
```

### Code Explanation

- `__len__()` is a dunder method.
- It tells Python what should happen when `len()` is used on a Team object.
- `len(team)` calls the `__len__()` method automatically.

## 23. super()

### What is super()?

`super()` is used to access methods or the constructor of a parent class from a child class.

It is useful when we want to reuse the parent's implementation.

### Example

A child class can use the parent constructor using `super()`.

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course


student = Student("Nandini", "AI/ML")

print("Name:", student.name)
print("Course:", student.course)
```

### Code Explanation

- `Person` is the parent class.
- `Student` inherits from `Person`.
- `super().__init__(name)` calls the constructor of the parent class.
- This allows the child class to reuse the parent's initialization logic.

## 24. __str__()

### What is __str__()?

`__str__()` defines a user-friendly string representation of an object.

It is automatically called when we use `print()` on an object.

### Example

We can use it to display useful information about a Student object.

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def __str__(self):
        return f"Student: {self.name}, Course: {self.course}"


student = Student("Nandini", "AI/ML")

print(student)
```

### Code Explanation

- `__str__()` returns a string representation of the object.
- When `print(student)` is executed, Python automatically calls `__str__()`.
- It makes the output more readable for users.

## 25. __repr__()

### What is __repr__()?

`__repr__()` provides a more detailed representation of an object.

It is mainly useful for debugging and development.

### Example

We can use it to show the important values stored inside an object.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student = Student("Nandini", 23)

print(repr(student))
```

### Code Explanation

- `__repr__()` defines a detailed representation of the object.
- `repr(student)` calls the `__repr__()` method.
- The output clearly shows the values stored in the object.
- It is especially useful while debugging.

## 26. __len__()

### What is __len__()?

`__len__()` is a special method that defines what should be returned when the `len()` function is used with an object.

### Example

We can use it to find the number of courses in a student's course list.

```python
class Student:
    def __init__(self, courses):
        self.courses = courses

    def __len__(self):
        return len(self.courses)


student = Student(["Python", "SQL", "Machine Learning"])

print("Number of courses:", len(student))
```

### Code Explanation

- `courses` stores a list of courses.
- `__len__()` returns the number of courses.
- `len(student)` automatically calls `__len__()`.
- This allows our custom object to work with Python's built-in `len()` function.

## 27. Dataclasses (Bonus)

### What is a Dataclass?

A dataclass is a special type of Python class that makes it easier to create classes mainly used for storing data.

The `@dataclass` decorator automatically provides methods such as `__init__()` and `__repr__()`.

This reduces the amount of code we need to write.

### Example

Instead of manually writing a constructor for a Student class, we can use a dataclass.

```python
from dataclasses import dataclass


@dataclass
class Student:
    name: str
    age: int
    course: str


student = Student("Nandini", 23, "AI/ML")

print(student)
```

### Code Explanation

- `dataclass` is imported from the `dataclasses` module.
- `@dataclass` is used above the `Student` class.
- `name`, `age`, and `course` are the fields of the class.
- We do not need to manually write the `__init__()` method.
- The dataclass automatically creates a useful `__repr__()` method.
- This makes the class shorter and easier to read.

## 28. Singleton Pattern (Bonus)

### What is a Singleton Pattern?

The Singleton Pattern is a design pattern that allows only one object of a class to be created.

If we try to create the class again, the same object is returned.

### Example

A configuration manager in an application may need only one shared object.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```

### Code Explanation

- `_instance` stores the single object created by the class.
- `__new__()` controls the creation of the object.
- If `_instance` is `None`, a new object is created.
- If an object already exists, the same object is returned.
- `object1 is object2` checks whether both variables refer to the same object.
- The output is `True`, which shows that only one instance is being used.

### Real-World Example

A configuration manager or logging manager can use the Singleton Pattern when the application needs one shared instance.
