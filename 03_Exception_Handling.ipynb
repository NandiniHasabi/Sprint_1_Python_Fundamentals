# Exception Handling

## Introduction

Exception handling is used to handle errors that occur while a Python program is running.

Normally, when an error occurs, the program stops. Exception handling allows us to handle the error properly and continue the program when possible.

Python provides the following keywords for exception handling:

- try
- except
- else
- finally
- raise
- assert

Python also allows us to create custom exceptions.

## 1. try

The `try` block contains code that may cause an exception.

Python first executes the code inside the `try` block.

If an error occurs, Python looks for a suitable `except` block to handle it.

### Example

We can put a division operation inside a `try` block because division by zero can cause an error.

```python
try:
    a = 10
    b = 0
    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Code Explanation

- The `try` block contains the division code.
- Since `number2` is 2, there is no error.
- The result is calculated and printed.
- If the denominator were 0, a `ZeroDivisionError` would occur.

## 2. except

The `except` block is used to handle an exception.

If an error occurs inside the `try` block, Python executes the matching `except` block.

### Example

Trying to divide a number by zero produces a `ZeroDivisionError`.

try:
    result = 10 / 0
    print(result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero")

### Code Explanation

- `10 / 0` causes a `ZeroDivisionError`.
- The `except ZeroDivisionError` block catches the error.
- Instead of stopping the program, it prints a useful error message.

## 3. else

The `else` block runs only when no exception occurs in the `try` block.

It is useful for code that should run only after the risky operation has completed successfully.

```python
try:
    number1 = 10
    number2 = 2
    result = number1 / number2

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Division successful")
    print("Result:", result)
```

### Code Explanation

- The division is performed inside `try`.
- No exception occurs.
- Therefore, the `else` block is executed.
- If division by zero occurred, the `else` block would not execute.

## 4. finally

The `finally` block always executes whether an exception occurs or not.

It is commonly used for cleanup operations such as closing files or database connections.

```python
try:
    number = 10
    result = number / 2
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("This block always executes")
```

### Code Explanation

- The `try` block performs the calculation.
- No exception occurs.
- The `finally` block still executes.
- The `finally` block executes even if an exception occurs.

## 5. try-except-else-finally

These four blocks can be used together.

- `try` → contains risky code
- `except` → handles an error
- `else` → runs when there is no error
- `finally` → always runs

```python
try:
    number1 = 20
    number2 = 4
    result = number1 / number2

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)

finally:
    print("Program execution completed")
```

## 6. ValueError

A `ValueError` occurs when a function receives a value of the correct type but the value is not valid for that operation.

### Example

Trying to convert the string `"hello"` into an integer causes a `ValueError`.

```python
try:
    age = int("hello")
    print(age)

except ValueError:
    print("Error: Please provide a valid integer")
```

### Code Explanation

- `int()` expects a value that can be converted into an integer.
- `"hello"` cannot be converted into an integer.
- Python raises `ValueError`.
- The `except` block handles the error.

## 7. TypeError

A `TypeError` occurs when an operation is performed using an inappropriate data type.

### Example

We cannot directly add a string and an integer.

```python
try:
    number = "10"
    result = number + 5
    print(result)

except TypeError:
    print("Error: Cannot add a string and an integer")
```

### Code Explanation

- `"10"` is a string.
- `5` is an integer.
- Python cannot directly add these two different types.
- Therefore, a `TypeError` occurs.
- The `except` block handles it.

## 8. IndexError

An `IndexError` occurs when we try to access an index that does not exist in a sequence such as a list or tuple.

```python
try:
    numbers = [10, 20, 30]
    print(numbers[5])

except IndexError:
    print("Error: The index does not exist")
```

### Code Explanation

- The list has indexes 0, 1, and 2.
- Index 5 does not exist.
- Therefore, Python raises `IndexError`.
- The `except` block handles it.

## 9. KeyError

A `KeyError` occurs when we try to access a dictionary using a key that does not exist.

```python
try:
    student = {
        "name": "Nandini",
        "age": 23
    }

    print(student["marks"])

except KeyError:
    print("Error: The key does not exist")
```

### Code Explanation

- The dictionary contains `name` and `age`.
- It does not contain the key `marks`.
- Therefore, Python raises `KeyError`.
- The exception is handled by the `except` block.

## 10. AttributeError

An `AttributeError` occurs when an object does not have the attribute or method that we try to access.

```python
try:
    number = 10
    number.append(5)

except AttributeError:
    print("Error: This object does not have the requested attribute or method")
```

## 11. NameError

A `NameError` occurs when we try to use a variable or name that has not been defined.

```python
try:
    print(student_name)

except NameError:
    print("Error: The variable has not been defined")
```

## 12. FileNotFoundError

A `FileNotFoundError` occurs when Python tries to open a file that does not exist at the specified location.

```python
try:
    with open("missing_file.txt", "r") as file:
        content = file.read()

except FileNotFoundError:
    print("Error: The requested file was not found")
```

## 13. ZeroDivisionError

A `ZeroDivisionError` occurs when we try to divide a number by zero.

```python
try:
    result = 100 / 0
    print(result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed")
```

## 14. ImportError

An `ImportError` occurs when Python cannot import a requested name from a module.

### Example

```python
try:
    from math import unknown_function

except ImportError:
    print("Error: The requested name could not be imported")
```

### Code Explanation

- Python tries to import `unknown_function` from the `math` module.
- That name does not exist in the module.
- Therefore, Python raises `ImportError`.
- The `except` block handles it.

## 15. ModuleNotFoundError

A `ModuleNotFoundError` occurs when Python cannot find the module that we are trying to import.

```python
try:
    import module_that_does_not_exist

except ModuleNotFoundError:
    print("Error: The module was not found")
```

## 16. raise

The `raise` keyword is used to manually generate an exception.

It is useful when we want to stop an operation because a condition is not valid.

```python
age = -5

try:
    if age < 0:
        raise ValueError("Age cannot be negative")

    print("Age:", age)

except ValueError as error:
    print("Error:", error)
```

### Code Explanation

- `age` contains `-5`.
- A negative age is not valid in this example.
- `raise ValueError()` manually creates an exception.
- The `except` block catches the exception.

## 17. assert

The `assert` statement is used to check whether a condition is true.

If the condition is false, Python raises an `AssertionError`.

It is commonly used for debugging and checking assumptions.

```python
age = 23

try:
    assert age >= 18, "Age must be 18 or above"
    print("Age requirement satisfied")

except AssertionError as error:
    print("Error:", error)
```

## 18. Multiple except Blocks

A program can have more than one `except` block.

Different types of exceptions can be handled separately.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ValueError:
    print("Please enter a valid number")

except ZeroDivisionError:
    print("Zero is not allowed")
```

### Code Explanation

- `ValueError` handles invalid input such as text.
- `ZeroDivisionError` handles the case where the user enters 0.
- Each exception has its own error message.

## 19. Custom Exceptions

A custom exception is an exception created by the programmer for a specific situation.

We create a custom exception by inheriting from the built-in `Exception` class.

### Example

We can create a custom exception for an invalid age.

```python
class InvalidAgeError(Exception):
    pass


age = 15

try:
    if age < 18:
        raise InvalidAgeError("Age must be 18 or above")

    print("Eligible")

except InvalidAgeError as error:
    print("Error:", error)
```

### Code Explanation

- `InvalidAgeError` is our custom exception.
- It inherits from Python's `Exception` class.
- We use `raise` to generate our custom exception.
- The `except` block catches the custom exception.

Custom exceptions are useful when a program needs to represent a specific business rule or application error.

## 20. Nested try-except

A `try-except` block can be placed inside another `try` block.

This is called nested exception handling.

It can be useful when different parts of a program need separate error handling.

```python
try:
    number = int(input("Enter a number: "))

    try:
        result = 100 / number
        print("Result:", result)

    except ZeroDivisionError:
        print("Cannot divide by zero")

except ValueError:
    print("Please enter a valid number")
```

### Code Explanation

- The outer `try` handles invalid input.
- The inner `try` handles division.
- `ValueError` is handled by the outer `except`.
- `ZeroDivisionError` is handled by the inner `except`.

## 21. Practical Example – Student Marks

We can use exception handling when accepting marks from a user.

The program should make sure that the input is a valid number and that the marks are within the valid range.

```python
try:
    marks = int(input("Enter your marks: "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100")

    print("Marks entered:", marks)

except ValueError as error:
    print("Invalid input:", error)
```

### Code Explanation

- The program asks the user to enter marks.
- `int()` converts the input into an integer.
- If the input is not a number, `ValueError` occurs.
- If the marks are below 0 or above 100, we manually raise `ValueError`.
- The `except` block handles both cases.
