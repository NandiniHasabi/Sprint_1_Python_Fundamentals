# Logging

Logging is the process of recording information about what happens while a program is running.

Logs can help us understand the program's execution, find errors, and monitor the application.

Instead of using `print()` everywhere, we can use Python's built-in `logging` module.

Logging is especially useful in large applications and production systems.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
logging.info("Processing the data")
logging.info("Program completed")
```

### Code Explanation

- `logging` is Python's built-in logging module.
- `logging.basicConfig()` sets the basic logging configuration.
- `level=logging.INFO` means INFO and higher-level messages will be shown.
- `logging.info()` records an informational message.

Logging helps us know what is happening inside a program.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in")
logging.info("Money deposited")
logging.warning("Low account balance")
```

## 2. Why Use Logging?

Logging is useful because it helps us:

- Track what a program is doing.
- Find and understand errors.
- Monitor applications.
- Store important information about program execution.
- Debug problems.
- Keep records of events.

### Example

Imagine a banking application.

Instead of using `print()` for every transaction, logging can record events such as:

- User logged in.
- Money was deposited.
- Money was withdrawn.
- An error occurred.

These records can help developers understand what happened in the application.

## 3. print() vs Logging

`print()` is mainly used to display information directly to the user.

Logging is designed to record information about a program and provides different levels such as INFO, WARNING, ERROR, and CRITICAL.

### Difference

| print() | Logging |
|---|---|
| Mainly displays information | Records program events |
| No log levels | Has different log levels |
| Not designed for application monitoring | Useful for monitoring |
| Usually appears only on the console | Can be saved to files |
| Limited configuration | Highly configurable |

For small programs, `print()` can be useful.

For larger applications, logging is generally more useful.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
```

### Simple Understanding

`print()` is like directly speaking to someone.

Logging is like maintaining a record or diary of what happened in the program.

## 4. Log Levels

Log levels tell us how important a log message is.

Python provides several commonly used log levels:

- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

The levels help developers understand the importance of a message.

## 5. DEBUG

`DEBUG` is used for detailed information that is mainly useful while developing or debugging a program.

It helps developers understand the internal working of the application.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("This is a debug message")
```

### Example

A developer may use DEBUG logs to check the value of a variable or to understand which step of a program is currently running.

## 6. INFO

`INFO` is used to record normal information about the program.

It tells us that something expected happened.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in successfully")
logging.info("Data processing started")
```

## 7. WARNING

`WARNING` indicates that something unexpected happened or that there may be a potential problem.

The program can usually continue running.

```python
import logging

logging.basicConfig(level=logging.WARNING)

logging.warning("Disk space is getting low")
```

### Example

A website may continue working even when the available storage space becomes low.

A WARNING message can inform the developer about the situation.

## 8. ERROR

`ERROR` indicates that a problem occurred and some part of the program could not perform an operation successfully.

```python
import logging

logging.basicConfig(level=logging.ERROR)

logging.error("Unable to connect to the database")
```

## 9. CRITICAL

`CRITICAL` represents a very serious problem that may prevent the application from continuing normally.

```python
import logging

logging.basicConfig(level=logging.CRITICAL)

logging.critical("Application cannot continue")
```

## 10. All Log Levels Together

The following example demonstrates all five common log levels.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug message")
logging.info("Info message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

### Important Point

The logging level controls which messages are displayed.

For example, if the level is set to `WARNING`, DEBUG and INFO messages will not normally be displayed.

WARNING, ERROR, and CRITICAL messages will be displayed.

## 11. Log Formatters

A log formatter controls how a log message is displayed.

We can include information such as:

- Date and time
- Log level
- Message

A formatter makes logs easier to read and understand.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("User logged in")
logging.warning("Password attempt failed")
```

### Code Explanation

- `%(asctime)s` shows the date and time.
- `%(levelname)s` shows the log level.
- `%(message)s` shows the actual log message.

The format makes each log entry easier to understand.

## 12. Console Handler

A console handler sends log messages to the console or terminal.

It is useful when we want to see logs while the program is running.

```python
import logging

logger = logging.getLogger("my_logger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

logger.addHandler(console_handler)

logger.info("This message is displayed in the console")
```

### Code Explanation

- `getLogger()` creates or gets a logger.
- `StreamHandler()` sends log messages to the console.
- `setLevel()` sets the minimum logging level.
- `logger.info()` creates an INFO message.

## 13. File Handler

A file handler saves log messages into a file.

This is useful when we want to keep a record of program activity and errors for later use.

```python
import logging

logger = logging.getLogger("file_logger")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")
file_handler.setLevel(logging.INFO)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

logger.info("Application started")
logger.warning("This is a warning")
logger.error("An error occurred")
```

### Code Explanation

- `FileHandler()` creates a handler that writes logs to a file.
- The file is named `application.log`.
- `Formatter()` defines the format of each log message.
- `setFormatter()` applies the format.
- The log messages are stored in the file.

After running this code, an `application.log` file will be created in the notebook's working directory.

## 14. Console and File Handlers Together

We can use both a console handler and a file handler.

This allows us to display logs on the screen and also save them to a file.

```python
import logging

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler = logging.StreamHandler()
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler("application.log")
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.info("Application started")
logger.warning("Low memory warning")
logger.error("Unable to process request")
```

### Code Explanation

The console handler displays the logs on the screen.

The file handler saves the same logs into `application.log`.

Using both handlers is useful when we want to monitor the application in real time and also keep a permanent record.

## 15. Rotating Logs (Bonus)

A log file can become very large if an application runs for a long time.

Rotating logs means creating a new log file when the current file reaches a particular size.

Python provides `RotatingFileHandler` for this purpose.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_logger")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "application_rotating.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for i in range(100):
    logger.info(f"Processing record {i}")

print("Rotating logs completed")
```

### Code Explanation

- `RotatingFileHandler` manages log file rotation.
- `maxBytes=1000` sets the approximate maximum size of the log file.
- `backupCount=3` keeps up to three backup log files.
- When the main log file becomes large, older logs can be moved to backup files.

This prevents a single log file from becoming too large.

## 16. Logging Best Practices

Some useful logging practices are:

1. Use the correct log level.
2. Use meaningful log messages.
3. Do not log sensitive information such as passwords.
4. Use logging instead of too many `print()` statements in production applications.
5. Use timestamps in important logs.
6. Save important logs to files when required.
7. Avoid logging unnecessary information.
8. Use clear and readable log messages.
9. Use rotating logs for applications that generate many logs.
10. Keep logs useful for debugging and monitoring.

## 17. Practical Example – Login System

Logging can be used to record successful logins, failed attempts, and errors.

The following example demonstrates how logging can be used in a simple login system.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

username = "Nandini"
password = "python123"

entered_username = input("Enter username: ")
entered_password = input("Enter password: ")

if entered_username == username and entered_password == password:
    logging.info("User logged in successfully")
    print("Login successful")

else:
    logging.warning("Login attempt failed")
    print("Invalid username or password")
```
