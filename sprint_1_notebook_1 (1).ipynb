# 7. Control Flow

Control flow means controlling the order in which statements in a Python program are executed.

Normally, Python executes statements from top to bottom. Control flow statements allow us to make decisions, repeat a block of code, or change the normal execution of a program.

The main control flow statements in Python are:
- if
- elif
- else
- for loop
- while loop
- break
- continue
- pass

### Example

Imagine a traffic signal. If the signal is green, we can go. If it is red, we must stop. The action depends on a condition.

Similarly, Python uses control flow statements to decide which code should execute.

```python
age = 20

if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
    
```

### Code Explanation

The variable 'age' stores the value 20.

The 'if' statement checks whether the age is greater than or equal to 18.

Since the condition is true, Python executes the 'print()' statement inside the `if` block.

If the condition were false, the 'else' block would execute.

## if Statement

The 'if' statement is used to execute a block of code only when a given condition is true.

If the condition is true, the code inside the 'if' block is executed. If the condition is false, it is skipped.

```python
temperature = 30

if temperature > 25:
    print("It is a hot day")
```

### Code Explanation

'temperature' stores the value 30.

The 'if' statement checks whether the temperature is greater than 25.

Since 30 is greater than 25, the message "It is a hot day" is printed.

## if-else Statement

The 'if-else' statement is used when we have two possible outcomes.

If the condition is true, the 'if' block is executed. Otherwise, the 'else' block is executed.

```python
number = 7

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

### Code Explanation

The '%' operator gives the remainder after division.

The program checks whether the remainder when 'number' is divided by 2 is zero.

If it is zero, the number is even. Otherwise, the number is odd.

## if-elif-else Statement

The 'if-elif-else' statement is used when we need to check multiple conditions.

Python first checks the 'if' condition. If it is false, it checks the 'elif' condition. If all conditions are false, the 'else' block is executed.

```python
marks = 75

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print("Grade:", grade)
```

### Code Explanation

The variable 'marks' stores 75.

Python first checks whether the marks are greater than or equal to 90. This condition is false.

Then it checks whether the marks are greater than or equal to 75. This condition is true, so 'grade' is assigned "B".

Python does not check the remaining conditions after finding a true condition.

## for Loop

A 'for' loop is used to repeat a block of code for each item in a sequence or collection.

It is useful when we know what items we want to iterate through.

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

### Code Explanation

The list 'fruits' contains three items.

The 'for' loop takes one item at a time from the list and stores it in the variable 'fruit'.

The 'print()' statement displays each fruit.

## while Loop

A 'while' loop repeatedly executes a block of code as long as a given condition is true.

It is useful when the number of repetitions depends on a condition.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### Code Explanation

The variable 'count' starts with the value 1.

The 'while' loop checks whether 'count' is less than or equal to 5.

The value is printed and then increased by 1 using 'count += 1'.

The loop stops when 'count' becomes 6 because the condition becomes false.

## break Statement

The 'break' statement is used to immediately stop a loop.

When Python reaches 'break', it exits the loop even if the loop condition is still true.

```python
for number in range(1, 10):
    if number == 5:
        break
    print(number)
```

### Code Explanation

The loop starts from 1 and goes up to 9.

When 'number' becomes 5, the 'if' condition becomes true.

The 'break' statement immediately stops the loop.

Therefore, the numbers printed are 1, 2, 3, and 4.

## continue Statement

The 'continue' statement skips the current iteration of a loop and moves to the next iteration.

It does not stop the complete loop.

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

### Code Explanation

The loop runs from 1 to 5.

When 'number' becomes 3, 'continue' skips that iteration.

The loop then continues with the next number.

Therefore, the output is:

1
2
4
5

## pass Statement

The 'pass' statement is a placeholder that does nothing when executed.

It is useful when we want to create a block of code but do not want to write its implementation yet.

```python
for number in range(1, 6):
    if number == 3:
        pass
    else:
        print(number)
```

### Code Explanation

When 'number' is 3, the 'pass' statement is executed.

'pass' does not perform any action, so nothing happens for that iteration.

The loop continues normally.

## List Comprehension

List comprehension is a simple and shorter way to create a new list from an existing sequence.

It allows us to write a loop and optionally a condition in a single line.

The basic syntax is:

[expression for item in iterable]

```python
numbers = [1, 2, 3, 4, 5]

squares = [number * number for number in numbers]

print(squares)
```

### Code Explanation

The list 'numbers' contains five numbers.

The list comprehension takes each number from the 'numbers' list and multiplies it by itself.

The expression 'number * number' calculates the square of each number.

The resulting values are stored in a new list called 'squares'.

List comprehension makes the code shorter compared to using a traditional 'for' loop.

## Dictionary Comprehension

Dictionary comprehension is a shorter and simpler way to create a dictionary using an iterable.

It allows us to create key-value pairs using a single line of code.

The basic syntax is:

{key: value for item in iterable}

### Example

Suppose we have a list of numbers:

[1, 2, 3, 4, 5]

We want to create a dictionary where each number is a key and its square is the value.

For example:

- 1 → 1
- 2 → 4
- 3 → 9
- 4 → 16
- 5 → 25

Dictionary comprehension allows us to create this dictionary in a short and readable way.

```python
numbers = [1, 2, 3, 4, 5]

squares = {number: number * number for number in numbers}

print(squares)
```

### Code Explanation

The list 'numbers' contains five numbers.

The dictionary comprehension takes each number from the list.

The number is used as the key, and 'number * number' is used as the value.

The resulting key-value pairs are stored in the 'squares' dictionary.

Dictionary comprehension makes the code shorter and easier to read compared to a traditional `for` loop.

## 8. Recursive Functions

A recursive function is a function that calls itself.

A recursive function must have a stopping condition. This is called the base condition.

### Example

We can use recursion to calculate the factorial of a number.

```python
def factorial(number):
    if number == 0:
        return 1

    return number * factorial(number - 1)

print(factorial(5))
```

### Explanation

- 'factorial()' is a recursive function because it calls itself.
- 'if number == 0' is the base condition.
- The function keeps calling itself with a smaller number.
- When the number becomes '0', the function stops.
- The factorial of 5 is '120'.

Calculation:

5 × 4 × 3 × 2 × 1 = 120

## 1. Iterators

An iterator is an object that gives us one value at a time.

We can use an iterator to go through the elements of a collection one by one.

Python provides two important functions for working with iterators:

- 'iter()' → creates an iterator
- 'next()' → gets the next value from the iterator

### Example

We can create an iterator from a list and get each value one by one.

```python
numbers = [10, 20, 30]

numbers_iterator = iter(numbers)

print(next(numbers_iterator))
print(next(numbers_iterator))
print(next(numbers_iterator))
```

### Explanation

- 'numbers' is a list containing three values.
- 'iter()' converts the list into an iterator.
- 'next()' gives the next value from the iterator.
- Each time we call 'next()', we get the next value.

The values are returned one at a time instead of getting all values at once.

## 2. Generators

A generator is a special type of function that produces values one at a time.

Generators are useful when we have a large amount of data because they do not store all the values in memory at the same time.

A generator function uses the `yield` keyword.

### Example

We can create a generator that produces numbers from 1 to 3.

```python
def numbers():
    yield 1
    yield 2
    yield 3

for number in numbers():
    print(number)
```

### Explanation

- 'numbers()' is a generator function.
- 'yield' produces one value at a time.
- The 'for' loop gets each value from the generator.
- The generator does not create all the values at the same time.

Generators are useful when working with large amounts of data.

## 3. yield

The 'yield' keyword is used in a generator function.

Unlike 'return', 'yield' does not completely stop the function. It pauses the function and remembers its current state.

When the function is called again, it continues from where it stopped.


```python
def count_numbers():
    yield 1
    yield 2
    yield 3

result = count_numbers()

print(next(result))
print(next(result))
print(next(result))
```

### Explanation

- 'yield 1' gives the first value.
- The function pauses after producing the value.
- The next 'next()' call continues the function.
- It then produces '2' and then '3'.

The main difference is:

'return' → returns a value and ends the function.

'yield' → returns a value and pauses the function.

## 4. Decorators

A decorator is used to add extra functionality to an existing function without changing the original function's code.

A decorator takes a function and adds some additional behavior to it.

### Example

We can create a decorator that prints a message before calling a function.

```python
def my_decorator(function):
    def wrapper():
        print("Starting the function")
        function()
    
    return wrapper


@my_decorator
def greet():
    print("Hello!")

greet()
```

### Explanation

- 'my_decorator' is a decorator function.
- It receives another function as input.
- 'wrapper()' adds extra behavior.
- '@my_decorator' applies the decorator to the 'greet()' function.
- When 'greet()' is called, the decorator runs first and then the original function runs.

Decorators allow us to add functionality without changing the original function.

## 5. Closures

A closure happens when an inner function remembers a value from the outer function even after the outer function has finished running.

In simple words, the inner function can remember information from its surrounding function.

### Example

We can create a function that remembers a person's name.

```python
def create_greeting(name):
    def greet():
        print("Hello", name)
    
    return greet


greeting = create_greeting("Nandini")

greeting()
```

### Explanation

- 'create_greeting()' is the outer function.
- 'greet()' is the inner function.
- The inner function uses the 'name' variable from the outer function.
- Even after 'create_greeting()' finishes, 'greet()' remembers the value of 'name'.
- This is called a closure.

## 6. global

The 'global' keyword is used when we want to change a variable that is created outside a function.

A variable created outside a function is called a global variable.

```python
count = 10

def update_count():
    global count
    count = 20

update_count()

print(count)
```

### Explanation

- 'count' is created outside the function, so it is a global variable.
- Normally, a function cannot directly change the global variable.
- The 'global' keyword tells Python that we want to use the global variable.
- The value of 'count' is changed from '10' to '20'.

## 7. nonlocal
The `nonlocal` keyword is used inside a nested function when we want to change a variable from the outer function.

It is mainly used with nested functions and closures.


```python
def counter():
    count = 0

    def increase():
        nonlocal count
        count = count + 1
        print(count)

    increase()
    increase()

counter()
```

### Explanation

- 'count' belongs to the outer 'counter()' function.
- 'increase()' is an inner function.
- 'nonlocal' allows the inner function to change the 'count' variable.
- Each time 'increase()' is called, the value of 'count' increases.
- The output is '1' and '2'.

The 'nonlocal' keyword is used for variables in an enclosing function.

## Why Do We Use Modules?

Modules are useful because they:

- Help us reuse code.
- Help us organize code.
- Reduce repeated code.
- Make programs easier to maintain.
- Provide ready-made functions that we can use.

## Types of Modules

There are two common types of modules:

1. Built-in Modules
2. User-defined Modules

### Built-in Modules

Built-in modules are already provided by Python.

Some examples are:

- `math` - used for mathematical operations
- `random` - used to generate random values
- `datetime` - used to work with dates and times
- `os` - used to work with the operating system

```python
import random

number = random.randint(1, 10)

print("Random number:", number)
```

### Explanation

- 'import random' imports the 'random' module.
- 'random.randint(1, 10)' generates a random integer between 1 and 10.
- The result can be different each time we run the program.

## User-Defined Module

A user-defined module is a Python file created by the programmer.

For example, we can create a file called `calculator.py` and add our own functions to it.

```python
# calculator.py

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
```

### Using the User-Defined Module

After creating `calculator.py`, we can import it into another Python program and use its functions.

```python
import calculator

print(calculator.add(10, 20))
print(calculator.multiply(5, 4))
```

### Explanation

- 'calculator.py' is our user-defined module.
- It contains the 'add()' and 'multiply()' functions.
- 'import calculator' imports the module.
- 'calculator.add()' calls the 'add()' function from the module.
- 'calculator.multiply()' calls the 'multiply()' function from the module.

This allows us to reuse functions without writing them again.

## Importing a Specific Function

We can import only the function that we need from a module.

This can make the code shorter and easier to read.

```python
from math import sqrt

print(sqrt(36))
```

### Explanation

- 'from math import sqrt' imports only the 'sqrt()' function from the 'math' module.
- We can use 'sqrt()' directly without writing 'math.sqrt()'.
- The output is '6.0'.

## Using an Alias

We can give a shorter name to a module using the `as` keyword.

This shorter name is called an alias.

```python
import math as m

print(m.sqrt(49))
```

### Explanation

- 'math' is imported with the alias 'm'.
- We can use 'm' instead of writing 'math'.
- 'm.sqrt(49)' finds the square root of 49.
- The output is '7.0'.

## What is a Package?

A package is a collection of related Python modules stored in a folder.

Packages help us organize multiple modules in a structured way.

For example, if a project has many Python files, we can group related files inside a package.

## Example

Suppose we have a project for a calculator.

We can organize it like this:

calculator_package/
    addition.py
    subtraction.py
    multiplication.py

Here, 'calculator_package' is the package and the Python files inside it are modules.

A package helps us keep related modules together.

## Creating a Package

A simple package can contain multiple Python files.

For example:

calculator_package/
    __init__.py
    addition.py
    multiplication.py

The '__init__.py' file tells Python that the folder can be treated as a package.

In modern Python versions, some packages can also work without '__init__.py', but it is commonly used for regular packages.

```python
# addition.py

def add(a, b):
    return a + b
```

## Importing Modules from a Package

We can import a module from a package and use its functions.

```python
from calculator_package import addition

print(addition.add(10, 20))
```

### Explanation

- 'calculator_package' is the package.
- 'addition' is a module inside the package.
- 'add()' is a function inside the 'addition' module.
- We import the 'addition' module from the package.
- Then we use 'addition.add()' to call the function.

## Importing a Function Directly

We can also import a specific function from a module inside a package.

```python
from calculator_package.addition import add

print(add(5, 10))
```

### Explanation

Here, we directly import the 'add()' function from the 'addition' module.

This allows us to call 'add()' directly without writing 'addition.add()'.

## Why Do We Use Packages?

Packages are useful because they:

- Organize related modules.
- Make large projects easier to manage.
- Help us reuse code.
- Make the project structure cleaner.
- Avoid keeping all code in one large Python file.

# Virtual Environments

## What is a Virtual Environment?

A virtual environment is an isolated environment used to install and manage Python packages for a specific project.

It keeps the packages of one project separate from the packages of other projects.

For example, one project may require TensorFlow 2.21, while another project may require a different version. Virtual environments help avoid conflicts between them.

## Why Do We Use Virtual Environments?

Virtual environments are useful because they:

- Keep project dependencies separate.
- Avoid conflicts between different package versions.
- Make projects easier to manage.
- Allow each project to have its own required packages.
- Help create a consistent development environment.

## Creating a Virtual Environment

We can create a virtual environment using the `venv` module provided by Python.

The following command creates a virtual environment named `myenv`.

```python
python -m venv myenv
```

## Activating a Virtual Environment

After creating the virtual environment, we need to activate it before installing packages.

### Windows

Use the following command:

myenv\Scripts\activate

### Linux or macOS

Use the following command:

'source myenv/bin/activate'

After activation, packages installed using 'pip' will be installed inside the virtual environment.

## Installing a Package

After activating the virtual environment, we can install Python packages using `pip`.

For example, we can install NumPy.

```python
pip install numpy
```

## Checking Installed Packages

We can use the following command to see the packages installed in the current environment.

```python
pip list
```

## Deactivating a Virtual Environment

After finishing our work, we can deactivate the virtual environment using:

```python
deactivate
```

## Example

Suppose I have two Python projects:

- Project A requires NumPy version 1.x.
- Project B requires a different NumPy version.

Installing both versions in the same environment can cause conflicts.

Instead, I can create two separate virtual environments:

Project A → Environment A → Required packages

Project B → Environment B → Required packages

This keeps the dependencies of both projects separate.

# Type Hinting 

## What is Type Hinting?

Type hinting is a way of specifying the expected data type of a variable, function parameter, or function return value.

It makes the code easier to understand and helps developers identify the expected type of data.

Python is still dynamically typed, so type hints do not force the type at runtime.

## Type Hinting for Variables

We can specify the expected type of a variable using a colon ':'.

### Example

Here, 'name' is expected to contain a string and 'age' is expected to contain an integer.

```python
name: str = "Nandini"
age: int = 23

print(name)
print(age)
```

### Explanation

- 'str' means the variable is expected to contain a string.
- 'int' means the variable is expected to contain an integer.
- Type hints make the expected data type clear to other developers.

## Type Hinting with Lists

We can also specify the type of elements expected inside a list.

For example, `list[int]` means a list containing integers.

```python
numbers: list[int] = [10, 20, 30, 40]

print(numbers)
```

# Built-in Functions in Python

## What are Built-in Functions?

Built-in functions are functions that are already provided by Python.

We can use them directly without defining them ourselves.

Some commonly used built-in functions are:

- 'map()'
- 'filter()'
- 'zip()'
- 'enumerate()'
- 'sorted()'
- 'any()'
- 'all()'
- 'isinstance()'
- 'id()'

These functions help us perform common operations easily and efficiently.

## 1. map()

The 'map()' function applies a function to every item in an iterable such as a list.

It is useful when we want to perform the same operation on multiple values.

### Example

We can use 'map()' to find the square of every number in a list.

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

### Explanation

- 'numbers' contains the values from 1 to 5.
- 'map()' applies the lambda function to every number.
- The lambda function calculates the square.
- 'list()' converts the result into a list.
- The output contains the squares of all the numbers.

## 2. filter()

### Concept

The 'filter()' function selects items from an iterable based on a condition.

It keeps only the values for which the condition is 'True'.

### Example

We can use 'filter()' to get only even numbers from a list.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

### Explanation

- 'filter()' checks each number using the given condition.
- 'x % 2 == 0' checks whether the number is even.
- Only numbers that satisfy the condition are selected.
- The output is '[2, 4, 6]'.

## 3. zip()

The `zip()` function combines elements from two or more iterables.

It pairs the elements based on their positions.

### Example

We can combine a list of names with a list of ages.

```python
names = ["Nandini", "Anu", "Priya"]
ages = [23, 22, 24]

students = list(zip(names, ages))

print(students)
```

### Explanation

- `zip()` combines the elements from both lists.
- The first name is paired with the first age.
- The second name is paired with the second age.
- The result contains pairs in the form of tuples.

## 4. enumerate()

The `enumerate()` function gives us both the index and the value while looping through an iterable.

It is useful when we need to know the position of each item.


```python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

### Explanation

- `enumerate()` provides both the index and the value.
- `index` stores the position.
- `fruit` stores the current item.
- By default, indexing starts from `0`.

Output:

0 Apple
1 Banana
2 Mango

## 5. sorted()

The `sorted()` function returns a new sorted list from an iterable.

By default, it sorts values in ascending order.

### Example

```python
numbers = [40, 10, 30, 20]

sorted_numbers = sorted(numbers)

print(sorted_numbers)
```

### Explanation

- `sorted()` arranges the values in ascending order.
- The original list is not changed.
- The result is a new sorted list.

Output:

[10, 20, 30, 40]

## 6. any()

The `any()` function returns `True` if at least one value in an iterable is `True`.

If all values are `False`, it returns `False`.


```python
values = [False, False, True, False]

result = any(values)

print(result)
```

### Explanation

- `any()` checks all the values.
- At least one value is `True`.
- Therefore, `any()` returns `True`.

We can remember it as:

At least one `True` → `True`
All `False` → `False`

## 7. all()

The `all()` function returns `True` only when all values in an iterable are `True`.

If even one value is `False`, it returns `False`.

### Example

```python
values = [True, True, True]

result = all(values)

print(result)
```

### Explanation

- `all()` checks every value.
- All the values are `True`.
- Therefore, the result is `True`.

We can remember it as:

All `True` → `True`
At least one `False` → `False`

## 8. isinstance()

The `isinstance()` function checks whether a value belongs to a particular data type or class.

It returns either `True` or `False`.

### Example

```python
age = 23

print(isinstance(age, int))
print(isinstance(age, str))
```

### Explanation

- `age` contains the integer value `23`.
- `isinstance(age, int)` checks whether `age` is an integer.
- It returns `True`.
- `isinstance(age, str)` checks whether `age` is a string.
- It returns `False`.

This function is useful when we need to check the type of a value.

## 9. id()

### Concept

The `id()` function returns the unique identity number of an object during its lifetime.

It can be used to identify whether two variables refer to the same object.

### Example

```python
name = "Nandini"

print(id(name))
```

### Explanation

- `name` refers to a string object.
- `id(name)` returns the identity of that object.
- The exact number can be different each time the program runs.
- The `id()` function is mainly useful when working with object identity.
